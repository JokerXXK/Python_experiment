```python
a = np.array([1,2,3,4,5,])
b = np.array(a)
print('a:', id(a),' b:', id(b))
print('='*40)
```

```python
b[0] = 10
print('a:', a,' b:', b)
print('='*40)
```

```python
a = np.array([1,2,3,4,5,])
b = np.array(a,copy=False)
print('a:', id(a),' b:', id(b))
print('='*40)
b[0] = 10
print('a:', a,' b:', b)
```

```python
a = np.array([1,2,3,4,5,])
print(a)

a = np.array([1,2,3,4,5,], ndmin = 2) 
a
print(a)
```

```python
a = np.mat([1,2,3,4])
print(type(a))

at = np.array(a, subok=True)
af = np.array(a, )
print('at, subok=True:', type(at))
print('af, subok=False:', type(af))
print(id(at),id(af))
```
