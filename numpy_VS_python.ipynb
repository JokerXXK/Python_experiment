numpy和原生python的性能测试

```python
import numpy as np
```

```python
np.__version__
```

```python
def python_sum(n):

    a = [i**2 for i in range(n)]
    b = [i**3 for i in range(n)]
    ab_sum = []
    for i in range(n):
        ab_sum.append(a[i] + b[i])
    return ab_sum
```

```python
python_sum(10)
```

```python
def numpy_sum(n):
    a = np.arange(n)**2
    b = np.arange(n)**3
    return a + b
```

```python
numpy_sum(10)
```

```python
%timeit python_sum(1000)
```

```python
%timeit numpy_sum(1000)
```

```python
%timeit python_sum(100000)
```

```python
%timeit numpy_sum(100000)
```

```python
%timeit python_sum(1000000)
```

```python
%timeit numpy_sum(1000000)
```
